## Tujuan
Mengimplementasikan A* pada graph kecil dan membandingkan beberapa heuristic.

```python
import heapq

graph = {
    'A': {'B': 2, 'C': 5},
    'B': {'C': 1, 'D': 4},
    'C': {'D': 1, 'E': 7},
    'D': {'E': 2},
    'E': {}
}
heuristic = {'A': 5, 'B': 4, 'C': 3, 'D': 2, 'E': 0}

def astar(graph, start, goal, h):
    frontier = [(h[start], 0, start, [start])]
    best_g = {start: 0}
    expanded = []
    while frontier:
        f, g, node, path = heapq.heappop(frontier)
        if g != best_g.get(node):
            continue
        expanded.append(node)
        if node == goal:
            return path, g, expanded
        for nxt, cost in graph[node].items():
            new_g = g + cost
            if new_g < best_g.get(nxt, float('inf')):
                best_g[nxt] = new_g
                heapq.heappush(frontier, (new_g + h[nxt], new_g, nxt, path + [nxt]))
    return None, float('inf'), expanded

path, cost, expanded = astar(graph, 'A', 'E', heuristic)
path, cost, expanded
```

### Eksperimen 1
Catat path, total cost, dan urutan node yang dieksplor. Jelaskan peran **g(n)**, **h(n)**, dan **f(n)=g(n)+h(n)**.

```python
h_zero = {k: 0 for k in heuristic}
for name, h in [('heuristic_awal', heuristic), ('h=0', h_zero)]:
    path, cost, expanded = astar(graph, 'A', 'E', h)
    print(name, '=>', path, 'cost=', cost, 'expanded=', expanded)
```

**Hasil Eksperimen 1:**
- Path: ['A', 'B', 'C', 'D', 'E']
- Total Cost: 5
- Expanded Nodes: ['A', 'B', 'C', 'D', 'E']

**Penjelasan Peran g(n), h(n), dan f(n):**
- **g(n)**: Biaya riil yang sudah ditempuh dari simpul awal (A) menuju simpul n.
- **h(n)**: Perkiraan/estimasi biaya dari simpul n menuju simpul tujuan (E).
- **f(n) = g(n) + h(n)**: Total perkiraan biaya jalur yang digunakan oleh priority queue untuk menentukan simpul mana yang di-expand berikutnya.

### Eksperimen 2
Bandingkan heuristic awal dengan **h=0**. Apakah solusi berubah? Apakah jumlah/urutan node yang dieksplor berubah?

**Hasil Perbandingan:**
- **Apakah solusi berubah?** Tidak. Jalur terbaik tetap `['A', 'B', 'C', 'D', 'E']` dengan total cost `5`.
- **Apakah urutan/jumlah node berubah?** Pada graf ini, urutan dan jumlah node yang dieksplorasi tidak berubah karena ukuran grafnya sangat kecil dan sederhana.
- **Catatan Konsep:** Ketika h = 0, fungsi $f(n) = g(n) + 0 = g(n)$, sehingga algoritma A* pada dasarnya berubah menjadi algoritma Dijkstra (Uniform Cost Search).
